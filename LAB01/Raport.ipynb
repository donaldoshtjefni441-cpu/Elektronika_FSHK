## Çfarë është Markdown?

Markdown është një mënyrë e thjeshtë për të formatuar tekstin. Përdoret në Jupyter Notebook për të shkruar shpjegime, për të vendosur tituj, lista, tabela dhe formula.

Me Markdown mund ta organizojmë më mirë një detyrë dhe ta bëjmë më të lehtë për t'u lexuar.



## Formatimi i tekstit

Në Markdown përdoren disa simbole për të ndryshuar pamjen e tekstit.

# Titull kryesor
## Nëntitull
### Titull më i vogël

**Tekst i trashë**

*Tekst i pjerrët*

~~Tekst i fshirë~~

Këto formate na ndihmojnë të dallojmë pjesët e rëndësishme të një shkrimi.

## Formulat dhe lidhjet

Në Jupyter Notebook mund të shkruajmë edhe formula matematikore.

Për shembull, ligji i Ohmit:

$$
U = I \cdot R
$$

Ku U është tensioni, I është rryma dhe R është rezistenca.
