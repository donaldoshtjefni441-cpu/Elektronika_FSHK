## 1. Qëllimi i eksperimentit

Në këtë laborator studiuam lidhjen e rezistencave në breadboard dhe mënyrën se si ndryshon tensioni në dalje të qarkut në varësi të rezistencave që përdorim.

Qëllimi ishte të krahasonim vlerat e matura në laborator me ato që përftohen nga formulat teorike.

## 2. Teoria

Rezistenca elektrike tregon kundërshtimin që i bëhet kalimit të rrymës elektrike. Ajo matet në Ohm (Ω).

Ligji i Ohmit jepet nga formula:

$$U=R\cdot I$$

Kur kemi dy rezistenca të lidhura në seri, tensionin në dalje e llogarisim me formulën:

$$V_{out}=V_{in}\cdot\frac{R_2}{R_1+R_2}$$

Për ndërtimin e qarkut përdorëm breadboard-in, i cili na mundëson lidhjen e komponentëve pa saldim.

## 4. Eksperimenti i parë – Rezistencat në kΩ

Lidhëm dy rezistenca në seri dhe matëm tensionin në hyrje dhe në dalje të qarkut.

| Madhësia | Vlera e matur |
|:---------|--------------:|
| R1 | 1.781 kΩ |
| R2 | 1.765 kΩ |
| Vin | 11.36 V |
| Vout | 5.68 V |

### Llogaritjet

Raporti i tensioneve të matura:

$$\frac{V_{out}}{V_{in}}=\frac{5.68}{11.36}=0.5$$

Raporti teorik:

$$\frac{R_2}{R_1+R_2}=\frac{1.765}{1.781+1.765}\approx0.498$$

Nga llogaritjet shohim se rezultatet janë shumë të afërta. Meqenëse rezistencat kanë pothuajse të njëjtën vlerë, tensioni në dalje është afërsisht gjysma e tensionit në hyrje.

---

## 5. Eksperimenti i dytë – Rezistencat në MΩ

Përsëritëm të njëjtën procedurë, por këtë herë përdorëm rezistenca me vlera më të mëdha.

| Madhësia | Vlera e matur |
|:---------|--------------:|
| R1 | 3.264 MΩ |
| R2 | 3.229 MΩ |
| Vin | 11.36 V |
| Vout | 4.83 V |

### Llogaritjet

Raporti i tensioneve:

$$\frac{4.83}{11.36}\approx0.425$$

Raporti teorik:

$$\frac{3.229}{3.264+3.229}\approx0.497$$

Në këtë rast kemi një diferencë më të madhe midis rezultatit teorik dhe atij të matur. Një nga arsyet mund të jetë rezistenca e brendshme e multimetrit, e cila ndikon në matje kur përdorim rezistenca në MΩ.

## 7. Përfundimi

Gjatë këtij laboratori mësuam si të lidhim rezistencat në breadboard dhe si të kryejmë matje me multimetër.

Në eksperimentin e parë rezultatet teorike dhe praktike ishin shumë të afërta, ndërsa në eksperimentin e dytë vumë re disa ndryshime.

Gjithashtu, nga eksperimenti i fundit kuptuam se ndryshimi i rezistencës së ngarkesës ndikon në tensionin në dalje të qarkut.
